```python
import os
import sys


import duckdb
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns


from sklearn.model_selection import train_test_split
```

```python
DATASET_DIR = "./datasets/"
DATALAKEHOUSE_DIR = "./datalakehouse/"

DUCKDB_FILE = os.path.join(DATALAKEHOUSE_DIR, "datalakehouse.duckdb")
```

```python
# load_datset

df = None
with duckdb.connect(DUCKDB_FILE) as conn: 
    df = conn.execute("SELECT * FROM s11_curated.student_performance_curated").df()
```

```python
df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=42)
df_train, df_valid = train_test_split(df_full_train, test_size=0.25, random_state=42)


df_test.to_parquet(os.path.join(DATASET_DIR, "student_performance_test.parquet"), index=False)
df_train.to_parquet(os.path.join(DATASET_DIR, "student_performance_train.parquet"), index=False)
df_valid.to_parquet(os.path.join(DATASET_DIR, "student_performance_valid.parquet"), index=False)
print("Datasets saved in parquet format:")
print(f"- {os.path.join(DATASET_DIR, 'student_performance_test.parquet')}")
print(f"- {os.path.join(DATASET_DIR, 'student_performance_train.parquet')}")
print(f"- {os.path.join(DATASET_DIR, 'student_performance_valid.parquet')}")
```
